# 12 - Employee Skills Table

**Objective:** Check whether the five raw datasets contain per-employee current skill information.

The professor specifically warns: "my current five datasets might not actually contain what skills each employee currently has."

```python
import pandas as pd

DATA_PATH = "../data/raw"

attrition = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")
engagement = pd.read_csv(f"{DATA_PATH}/hr_performance_engagement.csv")
essential = pd.read_csv(f"{DATA_PATH}/essential_skills.csv")
software = pd.read_csv(f"{DATA_PATH}/software_skills.csv")
occupation = pd.read_csv(f"{DATA_PATH}/occupation_data.csv")

print("Checking each dataset for employee-level skill data...")
print(f"  employee_attrition.csv: {attrition.shape[1]} columns")
print(f"  hr_performance_engagement.csv: {engagement.shape[1]} columns")
print(f"  essential_skills.csv: {essential.shape[1]} columns (role-level)")
print(f"  software_skills.csv: {software.shape[1]} columns (role-level)")
print(f"  occupation_data.csv: {occupation.shape[1]} columns (reference)")
```

```python
# Check attrition columns for any skill-related fields
skill_keywords = ['skill', 'competenc', 'ability', 'training', 'certif', 'proficien']
found = []
for col in attrition.columns:
    for kw in skill_keywords:
        if kw in col.lower():
            found.append(col)
            break

for col in engagement.columns:
    for kw in skill_keywords:
        if kw in col.lower():
            found.append(f"engagement:{col}")
            break

if found:
    print(f"Skill-related columns found: {found}")
else:
    print("No per-employee skill columns found in attrition or engagement datasets.")
```

```python
# essential_skills.csv contains ROLE-level skills (not per-employee)
print(f"essential_skills.csv: {len(essential)} rows, {essential['O*NET-SOC Code'].nunique()} unique roles")
print(f"  These are ROLE requirements, not individual employee skills")

print(f"\nsoftware_skills.csv: {len(software)} rows, {software['O*NET-SOC Code'].nunique()} unique roles")
print(f"  These are ROLE requirements, not individual employee skills")
```

## Conclusion

**The five raw datasets do NOT contain per-employee current skill information.**

- essential_skills.csv and software_skills.csv contain skill requirements at the **occupation/role level**
- employee_attrition.csv and hr_performance_engagement.csv have no skill-related columns

**Impact:** The skill gap engine handles this by accepting an empty employee_skills dictionary and computing gaps against role requirements.
