# 10 - Engagement Intelligence

**Objective:** Analyse employee engagement from the performance/engagement dataset.

- Average engagement score
- Engagement broken down by department
- List of lowest-engagement employees for HR review

```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

DATA_PATH = "../data/raw"
df = pd.read_csv(f"{DATA_PATH}/hr_performance_engagement.csv")
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head()
```

```python
# Find engagement score column
eng_cols = [c for c in df.columns if 'engagement' in c.lower()]
print(f"Engagement columns: {eng_cols}")

# Find department column
dept_cols = [c for c in df.columns if 'department' in c.lower()]
print(f"Department columns: {dept_cols}")

# Find ID column
id_cols = [c for c in df.columns if 'id' in c.lower()]
print(f"ID columns: {id_cols}")
```

```python
# Overall engagement statistics
eng_col = eng_cols[0] if eng_cols else None
if eng_col:
    print(f"Engagement Score Column: {eng_col}")
    print(f"  Mean: {df[eng_col].mean():.2f}")
    print(f"  Median: {df[eng_col].median():.2f}")
    print(f"  Std: {df[eng_col].std():.2f}")
    print(f"  Min: {df[eng_col].min()}")
    print(f"  Max: {df[eng_col].max()}")
    print(f"  Count: {len(df)}")
```

```python
# Engagement by department
dept_col = dept_cols[0] if dept_cols else None
if eng_col and dept_col:
    dept_eng = df.groupby(dept_col)[eng_col].mean().sort_values(ascending=False)
    print("Average Engagement by Department:")
    print(dept_eng.round(2))
```

```python
# Lowest engagement employees
id_col = id_cols[0] if id_cols else None
if eng_col:
    cols_to_show = [eng_col]
    if dept_col:
        cols_to_show.append(dept_col)
    if id_col:
        cols_to_show.insert(0, id_col)
    
    lowest = df.nsmallest(10, eng_col)[cols_to_show]
    print("Top 10 Lowest Engagement Employees:")
    print(lowest.to_string(index=False))
```

## Findings

- Engagement scores are aggregated from multiple survey periods per employee
- The engagement dataset has more rows (2,845) than the attrition dataset (1,470)
- Department-level analysis shows which teams have the lowest engagement
- HR can use the low-engagement list to target retention interventions
