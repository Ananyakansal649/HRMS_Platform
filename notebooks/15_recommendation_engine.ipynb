# 15 - Upskilling Recommendation Engine

**Objective:** Map missing skills to training recommendations.

Version 1 uses plain if/else rules mapping a missing skill to a training recommendation.

```python
import sys
sys.path.insert(0, '..')

from app.services.recommendation_service import recommend_for_skill, generate_recommendations

test_skills = ["MLOps", "Docker", "Python", "AWS", "Machine Learning", "Unknown Skill"]
for skill in test_skills:
    rec = recommend_for_skill(skill)
    print(f"  {skill}: {rec}")
```

```python
# Generate recommendations for employees
employee_roles = [
    {"employee_id": "301", "role": "Research Scientist"},
    {"employee_id": "302", "role": "Sales Executive"},
]
employee_skills = {
    "301": {"Python", "SQL"},
    "302": {"Excel"},
}

recs = generate_recommendations(employee_roles, employee_skills)
for r in recs:
    print(f"Employee {r['employee_id']} ({r['role']}):")
    print(f"  Gap count: {r['gap_count']}")
    for rec_text in r["recommendations"][:5]:
        print(f"    -> {rec_text}")
    print()
```
