## Objective

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

DATA_PATH = "../data/raw"
print("Files in data/raw:")
for f in sorted(os.listdir(DATA_PATH)):
    print(f"  - {f}")
```

## Dataset 1: employee_attrition.csv
**Expected purpose:** Attrition prediction (one row per employee, target = Attrition)
---

```python
df_attrition = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")

print("=== SHAPE ===")
print(f"Rows: {df_attrition.shape[0]}, Columns: {df_attrition.shape[1]}")
print()

print("=== COLUMN NAMES ===")
for i, col in enumerate(df_attrition.columns):
    print(f"  {i+1}. {col}")
print()

print("=== FIRST 5 ROWS ===")
df_attrition.head()
```

```python
print("=== DATA INFO ===")
df_attrition.info()
```

```python
print("=== MISSING VALUES (sorted) ===")
missing = df_attrition.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0] if missing.sum() > 0 else "No missing values found.")
print()
print(f"Total missing values: {df_attrition.isnull().sum().sum()}")
```

```python
print("=== DUPLICATE ROWS ===")
dup_count = df_attrition.duplicated().sum()
print(f"Duplicate rows: {dup_count}")
```

```python
print("=== POSSIBLE ID COLUMNS ===")
id_cols = [c for c in df_attrition.columns if "id" in c.lower()]
print(f"Columns with 'id' in name: {id_cols}")
print()

# Check uniqueness of candidate ID columns
for col in id_cols:
    unique_count = df_attrition[col].nunique()
    total = len(df_attrition)
    status = "(UNIQUE)" if unique_count == total else "(NOT UNIQUE)"
    print(f"  {col}: {unique_count} unique values out of {total} rows {status}")
```

```python
print("=== DATA TYPES ===")
print()
num_cols = df_attrition.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_attrition.select_dtypes(include=["object", "category", "str"]).columns.tolist()

print(f"Numerical columns ({len(num_cols)}):")
for c in num_cols:
    print(f"  - {c}")
print()
print(f"Categorical/text columns ({len(cat_cols)}):")
for c in cat_cols:
    print(f"  - {c}")
```

### Special: Attrition Target Distribution
The professor specifically requires checking the balance of the Attrition target variable.
---

```python
print("=== ATTRITION TARGET DISTRIBUTION ===")
print()
print("Value counts:")
print(df_attrition["Attrition"].value_counts())
print()
print("Percentage distribution:")
attrition_pct = df_attrition["Attrition"].value_counts(normalize=True) * 100
print(attrition_pct.round(2))
print()

yes_count = df_attrition["Attrition"].value_counts().get("Yes", 0)
no_count = df_attrition["Attrition"].value_counts().get("No", 0)
total = len(df_attrition)
print(f"Summary: Yes={yes_count} ({yes_count/total*100:.1f}%), No={no_count} ({no_count/total*100:.1f}%)")
print()
if yes_count / total < 0.2:
    print("WARNING: Attrition is imbalanced (< 20% Yes). Models will need stratification.")
```

## Data Understanding Summary

The table below consolidates what we found in each dataset.
---

```python
# Build summary table from actual data
datasets = {
    "employee_attrition": df_attrition,
    "hr_performance_engagement": df_engagement,
    "occupation_data": df_occupation,
    "essential_skills": df_essential,
    "software_skills": df_software,
}

purposes = {
    "employee_attrition": "Attrition prediction",
    "hr_performance_engagement": "Engagement/performance analytics",
    "occupation_data": "Occupation/role master",
    "essential_skills": "Essential skills required by roles",
    "software_skills": "Software/technology skills required by roles",
}

summary_rows = []
for name, df in datasets.items():
    rows, cols = df.shape
    missing = df.isnull().sum().sum()
    dups = df.duplicated().sum()
    id_candidates = [c for c in df.columns if "id" in c.lower() or "code" in c.lower()]
    id_info = "None found"
    if id_candidates:
        id_info = ", ".join(id_candidates)
    summary_rows.append({
        "Dataset": name,
        "Rows": rows,
        "Columns": cols,
        "Missing Values": int(missing),
        "Duplicate Rows": int(dups),
        "Possible ID": id_info,
        "Purpose": purposes[name],
    })

summary_df = pd.DataFrame(summary_rows)
summary_df
```

## Key Findings

### Employee Attrition
- The employee identifier column is **EmployeeNumber** (all values are unique).
- The Attrition column contains "Yes"/"No" values — it's the prediction target.
- `EmployeeCount` and `StandardHours` are constant across all rows — likely not useful for modeling.
- `Over18` is constant ("Y") — not useful.

### HR Performance Engagement
- This dataset uses "Employee ID" as identifier — we need to verify during Step 4 whether it aligns with EmployeeNumber in employee_attrition.
- Contains engagement scores, satisfaction scores, and work-life balance scores.

### Occupation Data
- Uses O*NET-SOC Code as the key — this is a standard occupation classification, not an employee ID.
- Contains role titles and descriptions — serves as a reference/lookup table.

### Essential Skills
- Also uses O*NET-SOC Code — links to occupation_data.
- Contains skill importance and level ratings per occupation.
- This is a many-to-one relationship (many skill rows per occupation).

### Software Skills
- Also uses O*NET-SOC Code — links to occupation_data.
- Contains software/technology tools required per occupation.
- Also many-to-one with occupation_data.
