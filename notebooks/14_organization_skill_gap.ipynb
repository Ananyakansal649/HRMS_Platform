# 14 - Organization-Wide Skill Gap

**Objective:** Roll up skill gaps across all employees.

Severity rule: 100+ missing -> HIGH, 50+ -> MEDIUM, else LOW.

```python
import pandas as pd
import sys
sys.path.insert(0, '..')

from app.services.skill_gap_service import get_organization_skill_gaps

DATA_PATH = "../data/raw"
attrition = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")

employee_roles = [
    {"employee_id": str(row["EmployeeNumber"]), "role": row["JobRole"]}
    for _, row in attrition.iterrows()
]
print(f"Employees: {len(employee_roles)}")
role_counts = pd.Series([e["role"] for e in employee_roles]).value_counts()
print(f"\nRole distribution:")
print(role_counts.to_string())
```

```python
# All employees start with no skills (raw data lacks per-employee skills)
employee_skills = {}

org_gaps = get_organization_skill_gaps(employee_roles, employee_skills)
print(f"Total skills with gaps: {len(org_gaps)}")

severity_counts = pd.Series([g["severity"] for g in org_gaps]).value_counts()
print(f"\nSeverity distribution:")
print(severity_counts.to_string())
```

```python
# Top 15 most critical organisation-wide skill gaps
print("Top 15 Critical Organisation Skill Gaps:")
print(f"{'Skill':<40} {'Missing':>8} {'Severity':<10}")
print("-" * 60)
for g in org_gaps[:15]:
    print(f"{g['skill']:<40} {g['missing_count']:>8} {g['severity']:<10}")
```

## Findings

- Organisation-wide gap analysis reveals which skills are most commonly missing
- Severity thresholds help HR prioritize training investments
- This data feeds the dashboard's skill gaps chart
