# 13 - Skill Gap Engine

**Objective:** Compute per-employee skill gaps using set subtraction.

`gap = required_skills - current_skills`

```python
import pandas as pd
import sys
sys.path.insert(0, '..')

from app.services.skill_gap_service import get_required_skills_by_role, compute_employee_skill_gaps

role_skills = get_required_skills_by_role()
print(f"Roles with defined skills: {len(role_skills)}")
print("\nSample roles:")
for role, skills in list(role_skills.items())[:3]:
    print(f"  {role}: {len(skills)} skills")
```

```python
# Compute skill gap for a single employee
employee_roles = [
    {"employee_id": "101", "role": "Research Scientist"},
]
employee_skills = {
    "101": {"Python", "SQL", "Machine Learning"},
}

gaps = compute_employee_skill_gaps(employee_roles, employee_skills)
for g in gaps:
    print(f"Employee {g['employee_id']} ({g['role']}):")
    print(f"  Required: {g['required_skills']} skills")
    print(f"  Has: {g['current_skills']} skills")
    print(f"  Gap: {g['gap_count']} skills missing")
    print(f"  Missing (first 5): {g['missing_skills'][:5]}")
```

```python
# Compare two employees with different skill levels
employee_roles = [
    {"employee_id": "201", "role": "Research Scientist"},
    {"employee_id": "202", "role": "Research Scientist"},
]
employee_skills = {
    "201": {"Python", "SQL", "Machine Learning", "Deep Learning", "TensorFlow"},
    "202": {"Python"},
}

gaps = compute_employee_skill_gaps(employee_roles, employee_skills)
for g in gaps:
    print(f"Employee {g['employee_id']}: {g['gap_count']} gaps")
    if g['gap_count'] > 0:
        print(f"  Top missing: {g['missing_skills'][:5]}")
```

## Findings

- The skill gap engine uses O*NET skill data to determine required skills per role
- Set subtraction gives the gap
- Employees with more existing skills have fewer gaps
- The engine handles fuzzy role name matching
